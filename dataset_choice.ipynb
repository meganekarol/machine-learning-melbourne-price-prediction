```python
import pandas as pd

def assess_messiness(df):
    print(f"Dataset shape: {df.shape}")
    print(f"Missing values: {df.isnull().sum().sum()} ({df.isnull().sum().sum() / df.size * 100:.1f}%)")
    print(f"Duplicate rows: {df.duplicated().sum()}")
    print(f"Data types: {df.dtypes.value_counts().to_dict()}")
    print(f"Columns with >50% missing: {(df.isnull().sum() > len(df) * 0.5).sum()}")
```

```python
df_red_wine = pd.read_csv("wine+quality/winequality-red.csv")  # Classification / Regression
df_white_wine = pd.read_csv("wine+quality/winequality-white.csv")
df_bank = pd.read_csv("bank.csv")  # Classification
df_melbourne = pd.read_csv("Melbourne_housing.csv", low_memory=False)  # Regression
```

```python
assess_messiness(df_red_wine)
```

```python
assess_messiness(df_white_wine)
```

```python
assess_messiness(df_melbourne)
```

```python
assess_messiness(df_bank)
```
